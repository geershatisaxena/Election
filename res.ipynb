#### ***CANDIDATE-SPECIFIC LOOKUPS***

```python
def find_candidate(first, last):
    """Helper: find rows where candidate name contains both tokens."""
    return df[
        df["Candidate"].str.contains(first, case=False, na=False)
        & df["Candidate"].str.contains(last, case=False, na=False)
    ]
```

```python
def constituency_contest(constituency, state):
    return df[(df["Constituency"] == constituency) & (df["State"] == state)] \
        .sort_values("Total Votes", ascending=False)
```

#### ***MARGIN / RUNNER-UP CALCULATIONS***

```python
df["Rank_in_seat"] = df.groupby(["State", "Constituency"])["Total Votes"] \
    .rank(ascending=False, method="first")
```

```python
winners = df[df["Rank_in_seat"] == 1].copy()
winners
```

```python
runners_up = df[df["Rank_in_seat"] == 2].copy()
runners_up
```

```python
merged = winners.merge(
    runners_up, on=["State", "Constituency"], suffixes=("_winner", "_runnerup")
)
merged["Margin"] = merged["Total Votes_winner"] - merged["Total Votes_runnerup"]
merged
```

```python
def animate_bar_race(categories, values, title, filename, horizontal=True,
                      color="steelblue", frames=16, fps=10, figsize=(9, 6), xlabel_rot=45):
    fig, ax = plt.subplots(figsize=figsize)
    values = np.array(values, dtype=float)
    def update(frame):
        ax.clear()
        frac = (frame + 1) / frames
        cur = values * frac
        if horizontal:
            ax.barh(categories, cur, color=color)
            ax.invert_yaxis()
            ax.set_xlim(0, values.max() * 1.1)
        else:
            ax.bar(categories, cur, color=color)
            ax.set_ylim(0, values.max() * 1.1)
            ax.tick_params(axis="x", rotation=xlabel_rot)
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_grouped_bar_race(categories, series_dict, title, filename, frames=16, fps=10, figsize=(10, 6)):
    fig, ax = plt.subplots(figsize=figsize)
    n = len(categories)
    width = 0.8 / len(series_dict)
    x = np.arange(n)
    colors = ["darkorange", "seagreen", "steelblue", "crimson"]
    maxval = max(max(v) for v in series_dict.values())
    def update(frame):
        ax.clear()
        frac = (frame + 1) / frames
        for i, (label, vals) in enumerate(series_dict.items()):
            cur = np.array(vals) * frac
            ax.bar(x + i * width, cur, width=width, label=label, color=colors[i % len(colors)])
        ax.set_xticks(x + width * (len(series_dict) - 1) / 2)
        ax.set_xticklabels(categories, rotation=45, ha="right")
        ax.set_ylim(0, maxval * 1.15)
        ax.set_title(title)
        ax.legend()
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_pie_pulse(labels, values, pulse_index, title, filename, frames=16, fps=10, figsize=(7, 7)):
    fig, ax = plt.subplots(figsize=figsize)
    def update(frame):
        ax.clear()
        t = frame / max(frames - 1, 1)
        explode = [0] * len(values)
        explode[pulse_index] = 0.04 + 0.14 * abs(np.sin(t * np.pi))
        ax.pie(values, labels=labels, autopct="%1.1f%%", explode=explode, startangle=90)
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_scatter_reveal(x, y, color_arr, title, filename, xlabel="", ylabel="",
                            frames=16, fps=12, figsize=(8, 6)):
    fig, ax = plt.subplots(figsize=figsize)
    x = np.array(x); y = np.array(y)
    n = len(x)
    idx_splits = np.linspace(0, n, frames + 1).astype(int)
    xmin, xmax = x.min(), x.max()
    ymin, ymax = y.min(), y.max()
    xr = (xmax - xmin) or 1
    yr = (ymax - ymin) or 1
    def update(frame):
        ax.clear()
        k = idx_splits[frame + 1]
        c = color_arr[:k] if hasattr(color_arr, "__len__") and not isinstance(color_arr, str) else color_arr
        ax.scatter(x[:k], y[:k], c=c, alpha=0.5, s=14)
        ax.set_xlim(xmin - 0.05 * xr, xmax + 0.05 * xr)
        ax.set_ylim(ymin - 0.05 * yr, ymax + 0.05 * yr)
        ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_bubble_growth(x, y, sizes, title, filename, frames=16, fps=10, figsize=(9, 7), color="mediumvioletred"):
    fig, ax = plt.subplots(figsize=figsize)
    x = np.array(x); y = np.array(y); sizes = np.array(sizes, dtype=float)
    def update(frame):
        ax.clear()
        frac = (frame + 1) / frames
        ax.scatter(x, y, s=sizes * frac, alpha=0.5, color=color)
        ax.set_xlim(0, x.max() * 1.1)
        ax.set_ylim(0, y.max() * 1.1)
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_kde_sequential(data_dict, title, filename, frames_per_curve=5, fps=10, figsize=(9, 6)):
    fig, ax = plt.subplots(figsize=figsize)
    items = list(data_dict.items())
    total_frames = frames_per_curve * len(items)
    def update(frame):
        ax.clear()
        shown = frame // frames_per_curve
        for label, vals in items[:shown + 1]:
            sns.kdeplot(vals, ax=ax, label=label, fill=True, alpha=0.2)
        ax.set_title(title)
        ax.legend()
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=total_frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_line_reveal(x, y, title, filename, xlabel="", ylabel="", frames=20, fps=14, figsize=(9, 6), fill=False):
    fig, ax = plt.subplots(figsize=figsize)
    x = np.array(x); y = np.array(y)
    n = len(x)
    idx_splits = np.linspace(1, n, frames).astype(int)
    def update(frame):
        ax.clear()
        k = idx_splits[frame]
        ax.plot(x[:k], y[:k], color="steelblue")
        if fill:
            ax.fill_between(x[:k], y[:k], alpha=0.3, color="steelblue")
        ax.set_xlim(x.min(), x.max())
        ax.set_ylim(y.min(), y.max() * 1.05 if y.max() > 0 else 1)
        ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_histogram_build(data, bins, title, filename, frames=14, fps=10, figsize=(9, 6), color="teal"):
    counts, edges = np.histogram(data, bins=bins)
    fig, ax = plt.subplots(figsize=figsize)
    def update(frame):
        ax.clear()
        frac = (frame + 1) / frames
        ax.bar(edges[:-1], counts * frac, width=np.diff(edges), align="edge", color=color)
        ax.set_ylim(0, counts.max() * 1.1)
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_donut_gauge(pct, title, filename, color="steelblue", frames=18, fps=12, figsize=(6, 6)):
    fig, ax = plt.subplots(figsize=figsize)
    def update(frame):
        ax.clear()
        frac = (frame + 1) / frames
        cur = pct * frac
        ax.pie([cur, 100 - cur], colors=[color, "#eeeeee"], startangle=90, counterclock=False,
               wedgeprops=dict(width=0.35))
        ax.text(0, 0, f"{cur:.1f}%", ha="center", va="center", fontsize=20)
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_radar_growth(categories, series, title, filename, frames=16, fps=10, figsize=(7, 7)):
    angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()
    angles += angles[:1]
    fig = plt.figure(figsize=figsize)
    ax = plt.subplot(111, polar=True)
    def update(frame):
        ax.clear()
        frac = (frame + 1) / frames
        for label, vals in series.items():
            v = [val * frac for val in vals] + [vals[0] * frac]
            ax.plot(angles, v, label=label)
            ax.fill(angles, v, alpha=0.1)
        ax.set_xticks(angles[:-1]); ax.set_xticklabels(categories)
        ax.set_title(title)
        ax.legend(loc="upper right", bbox_to_anchor=(1.35, 1.1), fontsize=7)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_stacked_bar_build(categories, seg1, seg2, seg1_label, seg2_label, title, filename,
                               frames=16, fps=10, figsize=(9, 6)):
    fig, ax = plt.subplots(figsize=figsize)
    seg1 = np.array(seg1, dtype=float); seg2 = np.array(seg2, dtype=float)
    def update(frame):
        ax.clear()
        frac = (frame + 1) / frames
        s1 = seg1 * frac; s2 = seg2 * frac
        ax.bar(categories, s1, color="royalblue", label=seg1_label)
        ax.bar(categories, s2, bottom=s1, color="gold", label=seg2_label)
        ax.set_ylim(0, (seg1.max() + seg2.max()) * 1.15)
        ax.set_title(title)
        ax.legend()
        plt.xticks(rotation=20, ha="right")
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_funnel_reveal(stages, values, title, filename, frames_per_stage=6, fps=10, figsize=(8, 6)):
    fig, ax = plt.subplots(figsize=figsize)
    total_frames = frames_per_stage * len(stages)
    values = np.array(values, dtype=float)
    def update(frame):
        ax.clear()
        stage_idx = frame // frames_per_stage
        stage_frac = (frame % frames_per_stage + 1) / frames_per_stage
        cur = []
        for i in range(len(stages)):
            if i < stage_idx:
                cur.append(values[i])
            elif i == stage_idx:
                cur.append(values[i] * stage_frac)
            else:
                cur.append(0)
        ax.barh(stages, cur, color="darkorange")
        ax.set_xlim(0, values.max() * 1.1)
        ax.invert_yaxis()
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=total_frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
def animate_area_reveal(x, y, title, filename, xlabel="", ylabel="", frames=18, fps=12, figsize=(9, 6), color="seagreen"):
    fig, ax = plt.subplots(figsize=figsize)
    x = np.array(x); y = np.array(y)
    n = len(x)
    idx_splits = np.linspace(1, n, frames).astype(int)
    def update(frame):
        ax.clear()
        k = idx_splits[frame]
        ax.plot(x[:k], y[:k], color=color)
        ax.fill_between(x[:k], y[:k], alpha=0.35, color=color)
        ax.set_xlim(x.min(), x.max())
        ax.set_ylim(0, y.max() * 1.05)
        ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.set_title(title)
        plt.tight_layout()
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
 
# ==============================================================================
# THE 50 ANIMATIONS
# ==============================================================================
 
# 1. Bar race — Top 15 Parties by Seats Won
top15 = seats_won.head(15)
animate_bar_race(top15.index.tolist(), top15.values, "Top 15 Parties by Seats Won",
                  "animations/01_top15_parties_seats.gif", horizontal=True, color="steelblue")
 
# 2. Bar race — Top 10 States by Number of Seats
top10_states = seats_per_state.head(10)
animate_bar_race(top10_states.index.tolist(), top10_states.values, "Top 10 States by Seat Count",
                  "animations/02_top10_states_seats.gif", horizontal=True, color="purple")
 
# 3. Bar race — Top 10 Parties by Total Votes
top10_votes = party_total_votes.head(10)
animate_bar_race(top10_votes.index.tolist(), top10_votes.values, "Top 10 Parties by Total Votes",
                  "animations/03_top10_parties_votes.gif", horizontal=True, color="darkorange")
 
# 4. Bar race — Top 10 Candidates by Total Votes Nationwide
top10_cand = df.nlargest(10, "Total Votes")
animate_bar_race(top10_cand["Candidate"].str.title().tolist(), top10_cand["Total Votes"].values,
                  "Top 10 Candidates by Total Votes", "animations/04_top10_candidates_votes.gif",
                  horizontal=True, color="seagreen")
 
# 5. Bar race — Top 10 Closest Contests (smallest margin)
close10 = merged.nsmallest(10, "Margin")
animate_bar_race(close10["Constituency"].tolist(), close10["Margin"].values,
                  "Top 10 Closest Contests (Margin)", "animations/05_closest_contests.gif",
                  horizontal=True, color="gold")
 
# 6. Bar race — Top 10 Largest Winning Margins
large10 = merged.nlargest(10, "Margin")
animate_bar_race(large10["Constituency"].tolist(), large10["Margin"].values,
                  "Top 10 Largest Winning Margins", "animations/06_largest_margins.gif",
                  horizontal=True, color="navy")
 
# 7. Bar race — Top 10 Parties by Strike Rate (Seats Won / Fielded)
top10_sr = strike_rate.head(10)
animate_bar_race(top10_sr.index.tolist(), top10_sr.values, "Top 10 Parties by Strike Rate (%)",
                  "animations/07_strike_rate.gif", horizontal=True, color="olive")
 
# 8. Bar race — Seats Won by Region
animate_bar_race(region_seats.index.tolist(), region_seats.values, "Seats Won by Region",
                  "animations/08_seats_by_region.gif", horizontal=False, color="teal", xlabel_rot=0)
 
# 9. Bar race — Top 10 Parties by Candidates Fielded
top10_fielded = candidates_fielded.head(10)
animate_bar_race(top10_fielded.index.tolist(), top10_fielded.values, "Top 10 Parties by Candidates Fielded",
                  "animations/09_candidates_fielded.gif", horizontal=True, color="slateblue")
 
# 10. Bar race — Top 10 Candidates by Postal Votes
top10_postal = df.nlargest(10, "Postal Votes")
animate_bar_race(top10_postal["Candidate"].str.title().tolist(), top10_postal["Postal Votes"].values,
                  "Top 10 Candidates by Postal Votes", "animations/10_top10_postal_votes.gif",
                  horizontal=True, color="crimson")
 
# 11. Grouped bar race — BJP vs INC Seats by Top 8 States
top8_states = seats_per_state.head(8).index
bjp_seats = winners[winners["Party"] == "Bharatiya Janata Party"].groupby("State").size().reindex(top8_states, fill_value=0)
inc_seats = winners[winners["Party"] == "Indian National Congress"].groupby("State").size().reindex(top8_states, fill_value=0)
animate_grouped_bar_race(top8_states.tolist(), {"BJP": bjp_seats.values, "INC": inc_seats.values},
                          "BJP vs INC Seats — Top 8 States", "animations/11_bjp_inc_grouped.gif")
 
# 12. Grouped bar race — BJP vs INC vs SP, Selected States
sel_states = ["Uttar Pradesh", "Gujarat", "Kerala", "Maharashtra", "West Bengal"]
sp_seats = winners[winners["Party"] == "Samajwadi Party"].groupby("State").size().reindex(sel_states, fill_value=0)
bjp_sel = winners[winners["Party"] == "Bharatiya Janata Party"].groupby("State").size().reindex(sel_states, fill_value=0)
inc_sel = winners[winners["Party"] == "Indian National Congress"].groupby("State").size().reindex(sel_states, fill_value=0)
animate_grouped_bar_race(sel_states, {"BJP": bjp_sel.values, "INC": inc_sel.values, "SP": sp_seats.values},
                          "BJP vs INC vs SP — Selected States", "animations/12_bjp_inc_sp_grouped.gif")
 
# 13. Pie pulse — Seat Share of Top 6 Parties (pulse #1 party)
top6 = seats_won.head(6)
others_sum = seats_won.iloc[6:].sum()
pie_labels = top6.index.tolist() + ["Others"]
pie_vals = top6.values.tolist() + [others_sum]
animate_pie_pulse(pie_labels, pie_vals, 0, "Seat Share — Top 6 Parties", "animations/13_seat_share_pulse.gif")
 
# 14. Pie pulse — Won vs Lost split (pulse Won)
wl_counts = df["Result"].value_counts()
animate_pie_pulse(wl_counts.index.tolist(), wl_counts.values, 0, "All Candidacies — Won vs Lost",
                   "animations/14_won_lost_pulse.gif")
 
# 15. Pie pulse — Seats by Region (pulse top region)
animate_pie_pulse(region_seats.index.tolist(), region_seats.values, 0, "Seat Share by Region",
                   "animations/15_region_pulse.gif")
 
# 16. Pie pulse — Candidate Type Breakdown (pulse Top 5 National Parties)
def classify(party):
    if party in top5_parties.tolist():
        return "Top 5 National Parties"
    if "independent" in party.lower():
        return "Independent"
    return "Other Party"
df["Candidate_type"] = df["Party"].apply(classify)
ctype = df["Candidate_type"].value_counts()
animate_pie_pulse(ctype.index.tolist(), ctype.values, 0, "All Candidacies by Type",
                   "animations/16_candidate_type_pulse.gif")
 
# 17. Pie pulse — Deposit Status (pulse Forfeited)
deposit_status = pd.Series({
    "Forfeited (<16.67%)": (df["% of Votes"] < 16.67).sum(),
    "Retained (>=16.67%)": (df["% of Votes"] >= 16.67).sum(),
})
animate_pie_pulse(deposit_status.index.tolist(), deposit_status.values, 0, "Deposit Status — All Candidates",
                   "animations/17_deposit_pulse.gif")
 
# 18. Pie pulse — BJP vs INC vs Others (pulse BJP)
winners["Party_group"] = winners["Party"].apply(
    lambda p: p if p in ["Bharatiya Janata Party", "Indian National Congress"] else "Others")
pg = winners["Party_group"].value_counts()
animate_pie_pulse(pg.index.tolist(), pg.values, 0, "BJP vs INC vs Others — Seats",
                   "animations/18_bjp_inc_others_pulse.gif")
 
# 19. Pie pulse — Top 6 Runner-up Parties (pulse #1)
ru6 = runners_up["Party"].value_counts().head(6)
animate_pie_pulse(ru6.index.tolist(), ru6.values, 0, "Top 6 Runner-up Parties",
                   "animations/19_runnerup_pulse.gif")
 
# 20. Pie pulse — Independent vs Party Candidacies (pulse Independent)
indep_split = pd.Series({
    "Independent": df["Party"].str.contains("Independent", case=False, na=False).sum(),
    "Party-affiliated": (~df["Party"].str.contains("Independent", case=False, na=False)).sum(),
})
animate_pie_pulse(indep_split.index.tolist(), indep_split.values, 0, "Independent vs Party-Affiliated Candidacies",
                   "animations/20_independent_pulse.gif")
 
# 21. Scatter reveal — Total Votes vs % of Votes, colored by Result (sampled 400 pts)
sample = df.sample(min(400, len(df)), random_state=42)
colors = sample["Result"].map({"Won": "green", "Lost": "red"}).values
animate_scatter_reveal(sample["Total Votes"].values, sample["% of Votes"].values, colors,
                        "Total Votes vs Vote % (green=Won, red=Lost)", "animations/21_votes_vs_pct_scatter.gif",
                        xlabel="Total Votes", ylabel="% of Votes")
 
# 22. Scatter reveal — Winner vs Runner-up Vote Share
animate_scatter_reveal(merged["% of Votes_winner"].values, merged["% of Votes_runnerup"].values, "purple",
                        "Winner vs Runner-up Vote Share — All Seats", "animations/22_winner_runnerup_scatter.gif",
                        xlabel="Winner's % of Votes", ylabel="Runner-up's % of Votes")
 
# 23. Scatter reveal — EVM vs Postal Votes (sampled)
sample2 = df.sample(min(400, len(df)), random_state=7)
animate_scatter_reveal(sample2["EVM Votes"].values, sample2["Postal Votes"].values, "darkgreen",
                        "EVM Votes vs Postal Votes (sampled)", "animations/23_evm_postal_scatter.gif",
                        xlabel="EVM Votes", ylabel="Postal Votes")
 
# 24. Scatter reveal — Margin vs Relative Margin %
animate_scatter_reveal(merged["Margin"].values, merged["Relative_margin_pct"].values, "firebrick",
                        "Margin vs Relative Margin (%)", "animations/24_margin_vs_relative.gif",
                        xlabel="Margin (votes)", ylabel="Relative Margin (%)")
 
# 25. Bubble growth — Seats vs Total Votes by State
bubble_df = pd.DataFrame({"seats": seats_per_state, "votes": votes_per_state}).dropna()
bubble_df["avg"] = bubble_df["votes"] / bubble_df["seats"]
animate_bubble_growth(bubble_df["seats"].values, bubble_df["votes"].values, bubble_df["avg"].values / 300,
                       "Seats vs Total Votes by State (bubble = avg votes/seat)", "animations/25_bubble_growth.gif")
 
# 26. KDE sequential build — BJP vs INC vs SP Vote Share Density
kde_data = {
    "BJP": winners[winners["Party"] == "Bharatiya Janata Party"]["% of Votes"].values,
    "INC": winners[winners["Party"] == "Indian National Congress"]["% of Votes"].values,
    "SP": winners[winners["Party"] == "Samajwadi Party"]["% of Votes"].values,
}
animate_kde_sequential(kde_data, "Winning Vote % Density — BJP vs INC vs SP", "animations/26_kde_sequential.gif")
 
# 27. Line reveal — Lorenz Curve of Winning Votes
sorted_votes = np.sort(winners["Total Votes"].values)
cum_votes = np.cumsum(sorted_votes) / sorted_votes.sum()
lorenz_x = np.linspace(0, 1, len(cum_votes))
animate_line_reveal(lorenz_x, cum_votes, "Lorenz Curve — Concentration of Winning Votes",
                     "animations/27_lorenz_curve.gif", xlabel="Cumulative Share of Seats",
                     ylabel="Cumulative Share of Votes", fill=True)
 
# 28. Line reveal — Cumulative National Vote Share by Party Rank
cum_share = (party_total_votes.sort_values(ascending=False).cumsum() / df["Total Votes"].sum() * 100)
animate_line_reveal(range(1, len(cum_share) + 1), cum_share.values, "Cumulative National Vote Share by Party Rank",
                     "animations/28_cumulative_vote_share.gif", xlabel="Party Rank", ylabel="Cumulative Vote Share (%)", fill=True)
 
# 29. Line reveal — Cumulative Seats by Party Rank (Pareto)
cum_seats = seats_won.sort_values(ascending=False).cumsum()
animate_line_reveal(range(1, len(cum_seats) + 1), cum_seats.values, "Cumulative Seats Across Parties (Pareto)",
                     "animations/29_pareto_seats.gif", xlabel="Party Rank", ylabel="Cumulative Seats")
 
# 30. Line reveal — Sorted Winning Vote Counts (all 543 seats)
sorted_winner_votes = sorted(winners["Total Votes"].values, reverse=True)
animate_line_reveal(range(len(sorted_winner_votes)), sorted_winner_votes, "Sorted Winning Vote Counts — All Seats",
                     "animations/30_sorted_winner_votes.gif", xlabel="Seat Rank", ylabel="Total Votes")
 
# 31. Line reveal — Candidate Efficiency Curve (cumulative fielded vs won)
eff = pd.DataFrame({"fielded": candidates_fielded, "won": seats_won}).fillna(0).sort_values("fielded", ascending=False)
animate_line_reveal(eff["fielded"].cumsum().values, eff["won"].cumsum().values,
                     "Candidate Efficiency Curve — All Parties", "animations/31_efficiency_curve.gif",
                     xlabel="Cumulative Candidates Fielded", ylabel="Cumulative Seats Won")
 
# 32. Histogram build — Winning Margin Distribution
animate_histogram_build(merged["Margin"].values, 30, "Winning Margin Distribution",
                         "animations/32_margin_hist_build.gif", color="darkred")
 
# 33. Histogram build — Winning Vote % Distribution
animate_histogram_build(winners["% of Votes"].values, 25, "Winning Vote % Distribution",
                         "animations/33_winpct_hist_build.gif", color="green")
 
# 34. Histogram build — EVM Votes Distribution
animate_histogram_build(df["EVM Votes"].values, 35, "EVM Votes Distribution — All Candidates",
                         "animations/34_evm_hist_build.gif", color="cornflowerblue")
 
# 35. Histogram build — Postal Votes Distribution
animate_histogram_build(df["Postal Votes"].values, 35, "Postal Votes Distribution — All Candidates",
                         "animations/35_postal_hist_build.gif", color="mediumpurple")
 
# 36. Donut gauge fill — Narendra Modi's Vote Share (Varanasi)
animate_donut_gauge(modi["% of Votes"], "Narendra Modi's Vote Share — Varanasi",
                     "animations/36_modi_gauge.gif", color="darkorange")
 
# 37. Donut gauge fill — Rahul Gandhi's Vote Share (Wayanad)
animate_donut_gauge(rahul_wayanad["% of Votes"], "Rahul Gandhi's Vote Share — Wayanad",
                     "animations/37_rahul_wayanad_gauge.gif", color="seagreen")
 
# 38. Donut gauge fill — Rahul Gandhi's Vote Share (Rae Bareli)
animate_donut_gauge(rahul_rb["% of Votes"], "Rahul Gandhi's Vote Share — Rae Bareli",
                     "animations/38_rahul_raebareli_gauge.gif", color="mediumseagreen")
 
# 39. Donut gauge fill — Akhilesh Yadav's Vote Share (Kannauj)
animate_donut_gauge(akhilesh["% of Votes"], "Akhilesh Yadav's Vote Share — Kannauj",
                     "animations/39_akhilesh_gauge.gif", color="crimson")
 
# 40. Donut gauge fill — Amit Shah's Vote Share (Gandhinagar)
animate_donut_gauge(amit_shah["% of Votes"], "Amit Shah's Vote Share — Gandhinagar",
                     "animations/40_amitshah_gauge.gif", color="steelblue")
 
# 41. Bar race — 4 Leaders' Total Votes Growing Together
animate_bar_race(leaders_labels, leaders["Total Votes"].values, "Total Votes — 4 Leaders",
                  "animations/41_leaders_votes_race.gif", horizontal=False, color="steelblue", xlabel_rot=0)
 
# 42. Bar race — 4 Leaders' Winning Margins Growing Together
animate_bar_race(leaders_labels, leaders_margin, "Winning Margins — 4 Leaders",
                  "animations/42_leaders_margin_race.gif", horizontal=False, color="crimson", xlabel_rot=0)
 
# 43. Stacked bar build — EVM + Postal Votes for the 4 Leaders
animate_stacked_bar_build(leaders_labels, leaders["EVM Votes"].values, leaders["Postal Votes"].values,
                           "EVM Votes", "Postal Votes", "EVM vs Postal Votes — 4 Leaders",
                           "animations/43_leaders_evm_postal_stack.gif")
 
# 44. Radar growth — 4 Leaders Normalized Comparison
radar_categories = ["Total Votes", "% of Votes", "Margin"]
radar_series = {}
tv_max = leaders["Total Votes"].max()
pct_max = leaders["% of Votes"].max()
mg_max = leaders_margin.max()
for i, lbl in enumerate(leaders_labels):
    radar_series[lbl.replace("\n", " ")] = [
        leaders["Total Votes"].values[i] / tv_max,
        leaders["% of Votes"].values[i] / pct_max,
        leaders_margin[i] / mg_max,
    ]
animate_radar_growth(radar_categories, radar_series, "Normalized Comparison — 4 Leaders",
                      "animations/44_leaders_radar_growth.gif")
 
# 45. Bar race — Top 10 Highest-Turnout Constituencies
seat_totals = df.groupby(["State", "Constituency"])["Total Votes"].sum().sort_values(ascending=False)
top10_turnout = seat_totals.head(10)
animate_bar_race([c for (_, c) in top10_turnout.index], top10_turnout.values,
                  "Top 10 Constituencies by Total Votes Cast", "animations/45_top10_turnout.gif",
                  horizontal=True, color="darkcyan")
 
# 46. Bar race — Close Contests (margin < 20,000) by Region
merged["Region"] = merged["State"].map(region_map)
close_by_region = merged[merged["Margin"] < 20000]["Region"].value_counts()
animate_bar_race(close_by_region.index.tolist(), close_by_region.values, "Close Contests (<20k Margin) by Region",
                  "animations/46_close_contests_region.gif", horizontal=False, color="darkred", xlabel_rot=0)
 
# 47. Bar race — Average Winning Margin by Region
avg_margin_region = merged.groupby("Region")["Margin"].mean().sort_values(ascending=False)
animate_bar_race(avg_margin_region.index.tolist(), avg_margin_region.values, "Average Winning Margin by Region",
                  "animations/47_avg_margin_region.gif", horizontal=False, color="cadetblue", xlabel_rot=0)
 
# 48. Funnel reveal — Candidates → Retained Deposit → Won
funnel_stages = ["Total Candidates", "Retained Deposit", "Won"]
funnel_values = [len(df), (df["% of Votes"] >= 16.67).sum(), len(winners)]
animate_funnel_reveal(funnel_stages, funnel_values, "Election Funnel — Candidates to Winners",
                      "animations/48_funnel_reveal.gif")
 
# 49. Area fill reveal — Cumulative Vote Share, Top 20 Parties
top20_share = (party_total_votes.head(20).cumsum() / df["Total Votes"].sum() * 100)
animate_area_reveal(range(1, len(top20_share) + 1), top20_share.values,
                     "Cumulative Vote Share — Top 20 Parties", "animations/49_area_cumulative_share.gif",
                     xlabel="Party Rank", ylabel="Cumulative Vote Share (%)")
 
# 50. Grand finale — 4-panel animated dashboard (all panels build together)
def animate_grand_finale(filename, frames=20, fps=10):
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    top5 = seats_won.head(5)
    modi_pct = modi["% of Votes"]
    lorenz_x_local = lorenz_x
    lorenz_y_local = cum_votes
    scatter_sample = df.sample(min(300, len(df)), random_state=1)
    sx = scatter_sample["Total Votes"].values
    sy = scatter_sample["% of Votes"].values
 
    def update(frame):
        frac = (frame + 1) / frames
 
        ax = axes[0, 0]; ax.clear()
        ax.bar(top5.index, top5.values * frac, color="steelblue")
        ax.set_title("Top 5 Parties by Seats")
        ax.tick_params(axis="x", rotation=30)
 
        ax = axes[0, 1]; ax.clear()
        cur_pct = modi_pct * frac
        ax.pie([cur_pct, 100 - cur_pct], colors=["darkorange", "#eeeeee"], startangle=90,
               wedgeprops=dict(width=0.35))
        ax.text(0, 0, f"{cur_pct:.1f}%", ha="center", va="center", fontsize=16)
        ax.set_title("Modi's Vote Share (Varanasi)")
 
        ax = axes[1, 0]; ax.clear()
        n = max(int(len(lorenz_x_local) * frac), 1)
        ax.plot(lorenz_x_local[:n], lorenz_y_local[:n], color="black")
        ax.plot([0, 1], [0, 1], linestyle="--", color="gray")
        ax.set_xlim(0, 1); ax.set_ylim(0, 1)
        ax.set_title("Lorenz Curve (Winning Votes)")
 
        ax = axes[1, 1]; ax.clear()
        k = max(int(len(sx) * frac), 1)
        ax.scatter(sx[:k], sy[:k], alpha=0.4, s=10, color="seagreen")
        ax.set_xlim(sx.min(), sx.max()); ax.set_ylim(sy.min(), sy.max())
        ax.set_title("Total Votes vs Vote % (build-up)")
 
        plt.suptitle("GE 2024 — Animated Highlights", fontsize=15)
        plt.tight_layout()
 
    ani = FuncAnimation(fig, update, frames=frames, repeat=False)
    ani.save(filename, writer=PillowWriter(fps=fps))
    plt.close(fig)
 
animate_grand_finale("animations/50_grand_finale_dashboard.gif")
```
