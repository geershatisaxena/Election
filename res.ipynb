#### ***CANDIDATE-SPECIFIC LOOKUPS***

```python
def find_candidate(first, last):
    """Helper: find rows where candidate name contains both tokens."""
    return df[
        df["Candidate"].str.contains(first, case=False, na=False)
        & df["Candidate"].str.contains(last, case=False, na=False)
    ]
```

```python
modi = find_candidate("NARENDRA", "MODI")
modi[["Constituency", "State", "Party", "Total Votes", "% of Votes", "Result"]]
```

```python
def constituency_contest(constituency, state):
    return df[(df["Constituency"] == constituency) & (df["State"] == state)] \
        .sort_values("Total Votes", ascending=False)
```

#### ***MARGIN / RUNNER-UP CALCULATIONS***

```python
df["Rank_in_seat"] = df.groupby(["State", "Constituency"])["Total Votes"] \
    .rank(ascending=False, method="first")
```

```python
winners = df[df["Rank_in_seat"] == 1].copy()
winners
```

```python
runners_up = df[df["Rank_in_seat"] == 2].copy()
runners_up
```

```python
merged = winners.merge(
    runners_up, on=["State", "Constituency"], suffixes=("_winner", "_runnerup")
)
merged["Margin"] = merged["Total Votes_winner"] - merged["Total Votes_runnerup"]
merged
```

```python
merged.nlargest(10, "Margin")[["Constituency", "State", "Candidate_winner", "Margin"]]
```

```python
merged.nsmallest(10, "Margin")[["Constituency", "State"
                                , "Candidate_winner", "Candidate_runnerup", "Margin"]]
```

```python
merged[merged["Candidate_winner"].str.contains("AMIT SHAH", case=False, na=False)][["Constituency", "Margin"]]
```

```python
merged[merged["Candidate_winner"].str.contains("NARENDRA MODI", case=False, na=False)][["Constituency", "Margin"]]
```

```python
merged.groupby("State")["Margin"].mean().sort_values(ascending=False)
```

```python
merged.groupby("Party_winner")["Margin"].mean().sort_values(ascending=False)
```
